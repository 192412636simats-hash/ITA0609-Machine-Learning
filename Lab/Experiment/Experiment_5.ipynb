```python
import numpy as np
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
```

```python
# Sample dataset

X = np.array([
    [1,2],
    [2,3],
    [3,3],
    [6,5],
    [7,7],
    [8,6]
])

# Classes
y = np.array([
    'Class A',
    'Class A',
    'Class A',
    'Class B',
    'Class B',
    'Class B'
])

print("Training Data:")
print(X)

print("Labels:")
print(y)
```

```python
# Create KNN classifier

knn = KNeighborsClassifier(n_neighbors=3)

# Train the model

knn.fit(X,y)

print("K-NN model trained successfully")
```

```python
# New sample for classification

test_data = np.array([[5,5]])

prediction = knn.predict(test_data)

print("Test Data:")
print(test_data)

print("Predicted Class:")
print(prediction)
```

```python
# Predict training data

y_pred = knn.predict(X)

accuracy = accuracy_score(y,y_pred)

print("Accuracy:")
print(accuracy*100,"%")
```
