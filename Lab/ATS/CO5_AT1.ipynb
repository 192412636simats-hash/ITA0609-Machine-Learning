```python
import pandas as pd
```

```python
transaction = pd.DataFrame({
    'Customer_ID': [101, 102, 103, 104, 105],
    'Transaction': [5000, 3000, 4500, 2000, 6000],
    'Product': ['Laptop', 'Phone', 'Tablet', 'Watch', 'Laptop']
})

transaction
```

```python
demographic = pd.DataFrame({
    'Customer_ID': [101, 102, 103, 104, 106],
    'Age': [25, 30, 28, 35, 40],
    'City': ['Chennai', 'Bangalore', 'Coimbatore', 'Chennai', 'Madurai']
})

demographic
```

```python
print("Duplicate transaction IDs:")
print(transaction[transaction['Customer_ID'].duplicated()])

print("Duplicate demographic IDs:")
print(demographic[demographic['Customer_ID'].duplicated()])
```

```python
combined = pd.merge(
    transaction,
    demographic,
    on='Customer_ID',
    how='outer'
)

combined
```

```python
print(combined.isnull().sum())
```

```python
transaction_ids = set(transaction['Customer_ID'])
demographic_ids = set(demographic['Customer_ID'])

print("Only in Transaction:", transaction_ids - demographic_ids)
print("Only in Demographic:", demographic_ids - transaction_ids)
```

```python
print(combined.groupby('Customer_ID').size())
```

```python
combined = combined.drop_duplicates(subset='Customer_ID')
```

```python
combined = combined.sort_values('Customer_ID')
```

```python
combined = combined.reset_index(drop=True)
```

```python
print("Final Combined Data:")
print(combined)
```
