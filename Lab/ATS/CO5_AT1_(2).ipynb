```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data1 = pd.DataFrame({
    'Customer_ID': [101, 102, 103, 104, 105],
    'City': ['Chennai', 'Bangalore', 'Chennai', 'Coimbatore', 'Bangalore'],
    'Product': ['Laptop', 'Phone', 'Laptop', 'Tablet', 'Phone'],
    'Sales': [50000, 30000, 45000, 20000, 35000]
})

data1.to_csv('sales1.csv', index=False)
```

```python
data2 = pd.DataFrame({
    'Customer_ID': [106, 107, 108, 109, 110],
    'City': ['Chennai', 'Bangalore', 'Coimbatore', 'Chennai', 'Bangalore'],
    'Product': ['Tablet', 'Laptop', 'Phone', 'Laptop', 'Tablet'],
    'Sales': [25000, 55000, 32000, 48000, 22000]
})

data2.to_csv('sales2.csv', index=False)
```

```python
df1 = pd.read_csv('sales1.csv')
df2 = pd.read_csv('sales2.csv')

print(df1)
print(df2)
```

```python
df = pd.concat([df1, df2], ignore_index=True)

print(df)
```

```python
print(df.info())
print(df.isnull().sum())
print(df.duplicated().sum())
```

```python
filtered = df[df['Sales'] > 30000]

print(filtered)
```

```python
city_sales = df.groupby('City')['Sales'].sum()

print(city_sales)
```

```python
sorted_data = df.sort_values('Sales', ascending=False)

print(sorted_data)
```

```python
print("Total Sales:", df['Sales'].sum())
print("Average Sales:", df['Sales'].mean())
print("Minimum Sales:", df['Sales'].min())
print("Maximum Sales:", df['Sales'].max())
```

```python
city_sales.plot(kind='bar')

plt.title('Sales by City')
plt.xlabel('City')
plt.ylabel('Total Sales')
plt.show()
```

```python
df['Sales'].plot(kind='hist', bins=5)

plt.title('Sales Distribution')
plt.xlabel('Sales')
plt.show()
```

```python
product_sales = df.groupby('Product')['Sales'].sum()

product_sales.plot(kind='pie', autopct='%1.1f%%')

plt.title('Sales by Product')
plt.ylabel('')
plt.show()
```

```python
plt.plot(df['Customer_ID'], df['Sales'], marker='o')

plt.title('Customer Sales')
plt.xlabel('Customer ID')
plt.ylabel('Sales')
plt.show()
```

```python
# Check missing values
print(df.isnull().sum())

# Check duplicate records
print("Duplicates:", df.duplicated().sum())

# Check data types
print(df.dtypes)

# Check invalid Sales values
df['Sales'] = pd.to_numeric(df['Sales'], errors='coerce')
print(df)
```
