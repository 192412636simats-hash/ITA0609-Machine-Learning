```python
import matplotlib.pyplot as plt
```

```python
plt.plot(df["Study_Hours"], df["Marks"], marker="o")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.title("Study Hours vs Marks")
plt.show()
```
