```python
import numpy as np
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import accuracy_score
```

```python
# XOR dataset

X = np.array([
    [0,0],
    [0,1],
    [1,0],
    [1,1]
])

y = np.array([
    0,
    1,
    1,
    0
])

print("Input Data:")
print(X)

print("Output:")
print(y)
```

```python
# Create Neural Network Model

model = MLPClassifier(
    hidden_layer_sizes=(2,),
    activation='logistic',
    solver='lbfgs',
    max_iter=1000
)

# Train the model
model.fit(X,y)

print("Training completed")
```

```python
# Prediction

prediction = model.predict(X)

print("Predicted Output:")
print(prediction)

# Accuracy
accuracy = accuracy_score(y,prediction)

print("Accuracy:")
print(accuracy*100,"%")
```

```python
test = np.array([[1,0]])

result = model.predict(test)

print("Test Input:")
print(test)

if result[0] == 1:
    print("Output: 1")
else:
    print("Output: 0")
```
