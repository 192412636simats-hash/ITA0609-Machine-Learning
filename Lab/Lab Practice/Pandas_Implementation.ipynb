```python
import pandas as pd

data = {
    'Student': ['Ravi', 'Priya', 'Kumar', 'Anu', 'Rahul', 'Sneha'],
    'Marks': [78, 92, None, 85, 88, None],
    'Attendance': [90, None, 85, 95, None, 80]
}

df = pd.DataFrame(data)

print("Student Data:")
print(df)

print("\nMissing Values:")
print(df.isnull())

print("\nCount of Missing Values:")
print(df.isnull().sum())
```

```python
import pandas as pd

data = {
    'Student': ['Ravi', 'Priya', 'Kumar', 'Anu', 'Rahul', 'Sneha'],
    'Marks': [78, 92, None, 85, 88, None],
    'Attendance': [90, None, 85, 95, None, 80]
}

df = pd.DataFrame(data)

print("Original Data:")
print(df)

df = df.dropna()

print("\nAfter Removing Missing Values:")
print(df)
```

```python
import pandas as pd

data = {
    'Student': ['Ravi', 'Priya', 'Kumar', 'Anu', 'Rahul', 'Sneha'],
    'Marks': [78, 92, None, 85, 88, None],
    'Attendance': [90, None, 85, 95, None, 80]
}

df = pd.DataFrame(data)

print("Original Data:")
print(df)

df['Marks'] = df['Marks'].fillna(df['Marks'].mean())
df['Attendance'] = df['Attendance'].fillna(df['Attendance'].mean())

print("\nAfter Filling Missing Values:")
print(df)
```

```python
import pandas as pd

data = {
    'Student': ['Ravi', 'Priya', 'Kumar', 'Anu', 'Rahul', 'Sneha'],
    'Department': ['ECE', 'CSE', 'EEE', 'ECE', 'CSE', 'EEE'],
    'Marks': [78, 92, 75, 85, 88, 80]
}

df = pd.DataFrame(data)

print("Student Data:")
print(df)

print("\nAverage Marks by Department:")
print(df.groupby('Department')['Marks'].mean())
```
